```python
import numpy as np
import cv2
from PIL import Image
```

```python
def preprocess_image(img):
    gray = cv2.cvtColor(np.array(img), cv2.COLOR_BGR2GRAY)
    resized = cv2.resize(gray, None, fx = 1.5, fy= 1.5, interpolation = cv2.INTER_LINEAR)
    processed_image = cv2.adaptiveThreshold(resized, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 61,11)
    return processed_image
```

```python
new_img= preprocess_image(pages[0])
Image.fromarray(new_img).show()
```

```python
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"
text = pytesseract.image_to_string(new_img)
```

```python
print(text)
```

```python
import re
pattern = r'Patient Information Birth Date(.*?)\(\d{3}\) '
matches = re.findall(pattern,text, re.DOTALL)
name_text  = matches[0].strip()
name_text
```

```python
pattern = r'((Jan|Feb|Mar|April|May|June|July|Aug|Sep|Oct|Nov|Dec)[ \d]++)'
date_matches = re.findall(pattern, name_text,  re.DOTALL)
date = date_matches[0][0]
```

```python
name_match = name_text.replace(date,'').strip()
name_match
```

```python
def remove_noise_from_name(name_text):
    pattern = r'((Jan|Feb|Mar|April|May|June|July|Aug|Sep|Oct|Nov|Dec)[ \d]++)'
    date_matches = re.findall(pattern, name_text, re.DOTALL)
    date = date_matches[0][0]
    name_match = name_text.replace(date,'').strip()
    return name_match
```

```python
name = remove_noise_from_name(name_text)
name
```
